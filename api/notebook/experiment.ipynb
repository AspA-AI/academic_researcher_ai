```python
def twoSum(nums, target):
    output = []
    for i in range(len(nums)-1):
        for j in range(1, len(nums)):
            print(i, j)
            if(nums[i]+ nums[j]==target):
                output.append(i)
                output.append(j)

    return output        

nums = [2,5,5,11]
target = 10
twoSum(nums, target)
```
